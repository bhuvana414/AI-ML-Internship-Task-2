```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
house_df = pd.read_csv("house_prices.csv")
house_df.head()
```

```python
house_df.isnull().sum()
```

```python
house_df.isnull().sum()[house_df.isnull().sum() > 0].plot(
    kind='bar',
    figsize=(10,5),
    title='House Prices - Missing Values'
)
plt.show()
```

```python
num_cols = house_df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = house_df.select_dtypes(include=['object']).columns
```

```python
for col in num_cols:
    house_df[col].fillna(house_df[col].median(), inplace=True)
```

```python
for col in cat_cols:
    house_df[col].fillna(house_df[col].mode()[0], inplace=True)
```

```python
threshold = len(house_df) * 0.5
house_df = house_df.dropna(thresh=threshold, axis=1)
```

```python
house_df.isnull().sum()
```

```python
house_df.to_csv("cleaned_house_prices.csv", index=False)
```

```python
medical_df = pd.read_csv("medical_appointments.csv")
medical_df.head()
```

```python
medical_df.isnull().sum()
```

```python
missing = medical_df.isnull().sum()
missing = missing[missing > 0]

if missing.empty:
    print("No missing values found in the Medical Appointment dataset.")
else:
    missing.plot(
        kind='bar',
        figsize=(10,5),
        title='Medical Appointments - Missing Values'
    )
    plt.show()
```

```python
num_cols = medical_df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = medical_df.select_dtypes(include=['object']).columns
```

```python
for col in num_cols:
    medical_df[col].fillna(medical_df[col].median(), inplace=True)
```

```python
for col in num_cols:
    medical_df[col].fillna(medical_df[col].median(), inplace=True)
```

```python
threshold = len(medical_df) * 0.5
medical_df = medical_df.dropna(thresh=threshold, axis=1)
```

```python
medical_df.isnull().sum()
```

```python
medical_df.to_csv("cleaned_medical_appointments.csv", index=False)
```
